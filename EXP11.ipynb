```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Create DataFrame
data = {
    "Name": ["Amit", "Riya", "John", "Sara", "Rahul"],
    "Marks": [85, 92, 78, 88, 95],
    "Age": [20, 19, 21, 20, 22]
}

df = pd.DataFrame(data)

# Show first 5 rows
df.head()
# Statistical summary
df.describe()

# Pie Chart (Marks distribution among students)

plt.figure(figsize=(6,6))

plt.pie(df["Marks"], labels=df["Name"], autopct="%1.1f%%", startangle=90)

# Title
plt.title("Student Marks Distribution (Pie Chart)")

# Equal aspect ratio makes pie circular
plt.axis("equal")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------
# Dataset (12 students)
# -------------------------
data = {
    "Name": ["Amit","Riya","John","Sara","Rahul","Ananya","Vikram","Neha","Karan","Priya","Aarav","Simran"],
    "Gender": ["M","F","M","F","M","F","M","F","M","F","M","F"],
    "Math": [45, 78, 62, 90, 55, 88, 33, 70, 99, 81, 60, 40],
    "Science": [50, 82, 65, 95, 58, 85, 35, 72, 98, 80, 63, 42],
    "Attendance": [80, 92, 75, 98, 70, 95, 60, 85, 99, 90, 78, 65]
}

df = pd.DataFrame(data)
df
df.head()
# Average marks (Math + Science)
df["Average"] = (df["Math"] + df["Science"]) / 2
df
df.describe()
df.loc[df["Average"].idxmax()], df.loc[df["Average"].idxmin()]
def grade(m):
    if m >= 90:
        return "A+"
    elif m >= 75:
        return "A"
    elif m >= 60:
        return "B"
    elif m >= 40:
        return "C"
    else:
        return "F"

df["Grade"] = df["Average"].apply(grade)
df
plt.figure(figsize=(10,5))  # wider figure

plt.bar(df["Name"], df["Attendance"], color="skyblue")

plt.title("Attendance Bar Graph")
plt.xlabel("Students")
plt.ylabel("Attendance %")

# rotate names so they don't overlap
plt.xticks(rotation=45)

plt.tight_layout()  # auto spacing fix

plt.show()
plt.figure(figsize=(6,4))
plt.plot(df["Name"], df["Average"], marker="o", color="green")
plt.title("Average Marks Trend")
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.show()
plt.figure(figsize=(10,5))

plt.plot(df["Name"], df["Average"], marker="o", color="green")

plt.title("Average Marks Line Graph")
plt.xlabel("Students")
plt.ylabel("Average Marks")


plt.xticks(rotation=45)

plt.grid(True)

plt.tight_layout()

plt.show()



plt.figure(figsize=(6,6))

# Convert average marks into categories (for meaningful pie chart)
def category(m):
    if m >= 80:
        return "Excellent"
    elif m >= 60:
        return "Good"
    elif m >= 40:
        return "Average"
    else:
        return "Poor"

df["Performance"] = df["Average"].apply(category)

# Count categories
perf_count = df["Performance"].value_counts()

plt.pie(
    perf_count,
    labels=perf_count.index,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0.05]*len(perf_count)
)

plt.title("Student Performance Distribution")
plt.axis("equal")

plt.tight_layout()
plt.show()
```
